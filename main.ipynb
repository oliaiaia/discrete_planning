# PART I

```python
import numpy as np
import matplotlib.pyplot as plt
from utils import plot_enviroment
import astar_planner

# Load the data
data = np.load('data_ps1.npz')
environment = data['environment']  
rod = data['rod'] 

print(f"Environment shape: {environment.shape}")
print(f"Rod shapes: {rod.shape}")
```

```python
# A: Visualize workspace and rod configurations
def part_A():
    fig, axes = plt.subplots(2, 2, figsize=(12, 12)) 
    
    orientations = [0, 45, 90, 135]
    
    # Plot rod configurations for each orientation
    for i, angle in enumerate(orientations):
        row = i // 2
        col = i % 2
        axes[row, col].imshow(rod[:, :, i], cmap='viridis', origin='lower')
        axes[row, col].set_title(f'Rod at {angle}°')
        axes[row, col].set_xlabel('X')
        axes[row, col].set_ylabel('Y')
    
    plt.tight_layout()
    plt.show()


    # # Plot workspace
    plt.figure(figsize=(20, 10))
    plt.imshow(environment, cmap='viridis', origin='lower')
    plt.title('\n Workspace \n')
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.grid(True)
    plt.show()
    

    print("- 4 orientations: 0°, 45°, 90°, 135°")
    print("- Provides coarse angular resolution (45° increments)")

part_A()
```

```python
# B: Visualize environment with object
def part_B():

    configs = [
        [90, 50, 0],   # (0=0°)
        [90, 50, 1],   # (1=45°)
        [90, 50, 2],   # (2=90°)
        [90, 50, 3]    # (3=135°)
    ]
    
    fig, axes = plt.subplots(2, 2, figsize=(12, 12))
    
    for i, config in enumerate(configs):
        row = i // 2
        col = i % 2
        merged_img = plot_enviroment(environment, rod, config)
        axes[row, col].imshow(merged_img, cmap='viridis', origin='lower')
        axes[row, col].set_title(f'Rod at ({config[0]},{config[1]}), orientation {config[2]}')
        axes[row, col].set_xlabel('X')
        axes[row, col].set_ylabel('Y')
    
    plt.tight_layout()
    plt.show()

part_B()
```

```python

def part_C_cpp():
    data = np.load('data_ps1.npz')
    environment, rod = data['environment'], data['rod']
    
    environment_list = environment.astype(int).tolist()
    rod_list = [rod[:, :, i].astype(int).tolist() for i in range(rod.shape[2])]
    
    cspace_list = astar_planner.compute_cspace(environment_list, rod_list)
    
    c_space = np.stack([np.array(mat) for mat in cspace_list], axis=2)
    
    fig, axes = plt.subplots(2, 2, figsize=(12, 12))
    orientations = [0, 45, 90, 135]
    
    for i, angle in enumerate(orientations):
        row = i // 2
        col = i % 2
        axes[row, col].imshow(c_space[:, :, i], cmap='viridis', origin='lower')
        axes[row, col].set_title(f'C-space at {angle}°')
        axes[row, col].set_xlabel('X')
        axes[row, col].set_ylabel('Y')
    
    plt.tight_layout()
    plt.show()
    
    return c_space

c_space = part_C_cpp()
```

```python
# D: Comment on obtained C-space
def part_D(c_space):
    print("\nD. Analysis of Configuration Space:")
    print("=" * 50)
    
    # Size of C-space
    print(f"C-space dimensions: {c_space.shape}")
    print(f"Total number of configurations: {np.prod(c_space.shape):,}")
    print(f"  - Spatial dimensions: 100 × 100 = 10,000 positions")
    print(f"  - Orientations: 4 discrete angles")
    print(f"  - Total: 100 × 100 × 4 = 400,000 configurations")
    
    # Analysis of free vs occupied space
    free_space = np.sum(c_space == 0)
    occupied_space = np.sum(c_space == 1)
    total_space = np.prod(c_space.shape)
    
    print(f"\nSpace distribution:")
    print(f"Free configurations: {free_space:,} ({free_space/total_space*100:.1f}%)")
    print(f"Occupied configurations: {occupied_space:,} ({occupied_space/total_space*100:.1f}%)")
    
    print(f"\nObservations:")
    print("1. The C-space expands the original 2D workspace to include orientation")
    print("2. Each orientation has different obstacle expansion patterns")
    print("3. Diagonal orientations (45°, 135°) typically have larger obstacle regions")
    print("4. The convolution method effectively 'grows' obstacles to account for robot size")
    print("5. C-space representation simplifies collision checking to point-in-space queries")
    
    print(f"\nPractical implications:")
    print("- Path planning must consider both position and orientation")
    print("- Some positions may be reachable only at certain orientations")
    print("- The discretization provides a searchable space for algorithms like A*")

part_D(c_space)
```

# PART II

```python
start_x, start_y, start_theta = 6, 6, 2
goal_x, goal_y, goal_theta = 85, 85, 0
```

```python
def plot_path(environment, path, title="Path"):
    plt.figure(figsize=(12, 10))
    
    plt.imshow(environment, cmap='viridis', origin='lower', alpha=0.7)
    
    if len(path) > 0:
        path_array = np.array(path)
        plt.plot(path_array[:, 0], path_array[:, 1], c='black', linewidth=2, label='Path')
        plt.scatter([start_x], [start_y], c='black', s=50, marker='o', label='Start')
        plt.scatter([goal_x], [goal_y], c='black', s=50, marker='s', label='Goal')
        
    plt.title(title)
    plt.xlabel('X')
    plt.ylabel('Y')
    plt.legend()
    plt.grid(True, alpha=0.3)
    plt.show()
```

```python
# to C++ format
environment_list = environment.astype(int).tolist()
rod_list = [rod[:, :, i].astype(int).tolist() for i in range(rod.shape[2])]

#  zero heuristic
planner_zero = astar_planner.AStar(environment_list, rod_list, "dijkstra")


path_zero = planner_zero.find_path(start_x, start_y, start_theta, goal_x, goal_y, goal_theta)

plot_path(environment, path_zero, "Path with dijkstra")


if len(path_zero) > 0:
    from utils import plotting_results
    plan_for_animation = [(step[1], step[0], step[2]) for step in path_zero]
    plotting_results(environment, rod, plan_for_animation, 'rod_solve.gif')
```

```python
# to C++ format
environment_list = environment.astype(int).tolist()
rod_list = [rod[:, :, i].astype(int).tolist() for i in range(rod.shape[2])]

#  zero heuristic
planner_zero = astar_planner.AStar(environment_list, rod_list, "manhattan")

start_x, start_y, start_theta = 6, 6, 2
goal_x, goal_y, goal_theta = 85, 85, 0

path_zero = planner_zero.find_path(start_x, start_y, start_theta, goal_x, goal_y, goal_theta)

plot_path(environment, path_zero, "Path with manhattan heuristic")
```
